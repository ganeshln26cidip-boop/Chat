# 3. Loop control: `break`, `continue`, and `pass`

```python
# break: stop the loop when i reaches 6
for i in range(1, 11):
    if i == 6:
        break
    print(i)
```

```python
# continue: skip even numbers
for i in range(1, 6):
    if i % 2 == 0:
        continue
    print(i)
```

```python
# pass: placeholder statement
for i in range(3):
    pass
print('Loop completed')
```
