# 8. Classroom challenges

Try these beginner exercises in Colab.

```python
# 1. Print numbers from 1 to 10
for i in range(1, 11):
    print(i)
```

```python
# 2. Print odd numbers between 1 and 50
for i in range(1, 51):
    if i % 2 != 0:
        print(i)
```

```python
# 3. Multiplication table for a user-entered number
num = int(input('Enter a number: '))
for i in range(1, 11):
    print(num, 'x', i, '=', num * i)
```

```python
# 4. Count vowels
text = 'Programming in Python'
count = 0
for ch in text.lower():
    if ch in 'aeiou':
        count += 1
print('Vowels =', count)
```

```python
# 5. Five-row star triangle
for row in range(1, 6):
    print('*' * row)
```
